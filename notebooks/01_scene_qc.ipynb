# Scene QC

Exploratory check for one cached Sentinel-2 folder. This notebook is not on the demo path.
Point `SCENE` at `data/scenes/{water_body_id}/{YYYYMMDD}` and confirm B02, B03, B04, B08, B11, B12, and SCL open, and that SCL cloud classes are not the majority of the AOI.

```python
from pathlib import Path
import numpy as np
import rasterio

SCENE = Path('../data/scenes/demo-reservoir/20241202')
needed = ['B02.tif', 'B03.tif', 'B04.tif', 'B08.tif', 'B11.tif', 'B12.tif', 'SCL.tif']
for name in needed:
    path = SCENE / name
    print(name, path.is_file())
with rasterio.open(SCENE / 'SCL.tif') as src:
    scl = src.read(1)
print('SCL unique', np.unique(scl, return_counts=True))
```
